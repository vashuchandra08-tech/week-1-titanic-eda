```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage': missing_percentage
})

missing_summary.sort_values(by='Missing Values', ascending=False)
```

```python
df = df.drop(columns=['Cabin'])
```

```python
df
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.dtypes
```

```python
print("Sex:", df['Sex'].unique())
print("Embarked:", df['Embarked'].unique())
print("Pclass:", df['Pclass'].unique())
print("Survived:", df['Survived'].unique())
```

```python
df.describe()
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='Survived')

plt.title('Passenger Survival Distribution')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x='Sex', hue='Survived')

plt.title('Survival Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Age', bins=30, kde=True)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = df.select_dtypes(include='number').corr()

sns.heatmap(correlation, annot=True)

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.boxplot(data=df, x='Fare')

plt.title('Fare Distribution and Outliers')
plt.xlabel('Fare')

plt.show()
```

```python
survival_rate = df['Survived'].mean() * 100
print(f"Overall survival rate: {survival_rate:.2f}%")
```

```python
gender_survival = df.groupby('Sex')['Survived'].mean() * 100
print(gender_survival)
```

```python
class_survival = df.groupby('Pclass')['Survived'].mean() * 100
print(class_survival)
```

```python
age_by_survival = df.groupby('Survived')['Age'].mean()
print(age_by_survival)
```

```python
fare_by_class = df.groupby('Pclass')['Fare'].mean()
print(fare_by_class)
```

```python
correlation_with_survival = (
    df.select_dtypes(include='number')
      .corr()['Survived']
      .sort_values(ascending=False)
)

print(correlation_with_survival)
```

```python
print("Final dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Total duplicate rows:", df.duplicated().sum())
```
